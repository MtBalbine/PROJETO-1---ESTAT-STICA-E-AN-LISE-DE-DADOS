1. Importar as bibliotecas Pandas (pd) e math

```python
import pandas as pd
import math
import matplotlib.pyplot as plt
import seaborn as sns
```

2. Definir a variável "arquivo" pelo nome da planilha.


```python
arquivo='economia_circular.xlsx'
```

3. Ler a planilha no Python. Escolha um nome para armazenar. Escolheremos df

```python
df=pd.read_excel(arquivo)
df
```

4.Agrupando os valores qualitativos a partir da variável selecionada.Escolhemos Escolaridade (nome da coluna no arquivo df). **Os valores serão armamzenados em df2.**

```python
df2=df.Porte.value_counts()
df2
```

5. Ordenando os valores da variável em ordem alfabética

```python
df2=df.Porte.value_counts().sort_index()
df2
```

6. Contando o total de elementos

```python
totaldf2=df2.sum()
totaldf2
```

7. Montando a tabela de frequências da variável



7.1. Criar um DataFrame para a coluna desejada (dadosvquali no exemplo)

```python
# Cria um DataFrame a partir da Série df2
dadosvquali = pd.DataFrame(df2).reset_index()

# Renomeia as colunas para algo mais descritivo
dadosvquali.columns = ['Porte', 'Frequência']

# Exibe o DataFrame
print(dadosvquali)
```

7.2 Calcular a frequencia relativa em % com 2 casas decimais

```python
# Calcula a porcentagem de frequência de cada categoria
dadosvquali['Frequência Relativa (%)'] = round(dadosvquali['Frequência'] * 100 / totaldf2, 2)

# Exibe o DataFrame atualizado
print(dadosvquali)
```

7.3 Criando uma nova tabela, renomeando as colunas

```python
tabela=pd.DataFrame({"Porte":dadosvquali['Porte'],"Frequência":dadosvquali['Frequência'],"Frequência Relativa (%)":dadosvquali["Frequência Relativa (%)"]})
```

```python
tabela
```

```python
rótulos1 = tabela['Porte']
valores1 = tabela['Frequência']
plt.figure(figsize=(7,7))
plt.pie(valores1,labels=rótulos1,autopct='%1.1f%%',startangle=90)
plt.legend()
plt.show()
```
